## Some ways to use the module to play around with complex numbers

This package is created so that you can toy around with complex numbers. Let us first import the complex number class from complex_lib, and some other useful packages:

```python
from complex_lib.complex import Complex
import numpy as np
```

### Let us create a complex number!

```python
# Create a complex number by first specifying real part and then imaginary part:
my_complex_number = Complex(1, 2)

# You can also create complex numbers using polar form, by specifying magnitude and angle
# Keep in mind that cartesian form is the default:
my_complex_number_polar = Complex.from_polar(1, np.pi/2)

# Let us print our number:
print(f"My first complex number: {my_complex_number}")
```

### Class operations:
You can perform different operations on the complex number class:

```python
z1 = Complex(1, 2)
z2 = Complex(3, 2)

print(f"My two complex numbers: {z1} and {z2}")

# You can add two numbers together:
print(f"z1 + z2 = {z1 + z2}")

# Subtract:
print(f"z1 - z2 = {z1 - z2}")

# Scalar multiplication:
print(f"2 * z1 = {2 * z1}")

# Multiplication:
print(f"z1 * z2 = {z1 * z2}")

# Division, both with two complex number and with real numbers:
print(f"z1 / z2 = {z1 / z2}")

print(f"1 / z1 = {1 / z1}")
```

## Calculate roots:
You can also use the complex class to both calculate roots of a complex number and plot them!

```python
# Create a complex number:
z = Complex.from_polar(1, np.pi/3)

# Calculate its roots
z_roots = z.calculate_roots(degree = 3)

print(f"The roots of {z} are:")

for root in z_roots:
    print(f"{root}")
```

```python
# On to plotting!
# Given our roots z_roots we can easily plot:

z.plot_roots(roots = z_roots)
```
